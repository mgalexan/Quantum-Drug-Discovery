```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots

plt.style.use("science")
```

```python
# load LV data
lv_data_numerical = np.load("results/lk-final_numerical.npz")
lv_data_quantum = np.load("results/lk-final_quantum.npz")
```

```python
t = lv_data_numerical["t"]
numerical = lv_data_numerical["x"]
quantum = lv_data_quantum["x"]
```

```python
# compare the quantum/numerical trajectories and phase space
fig, anx = plt.subplots(1, 2, figsize=(8, 3))
anx[0].plot(t, numerical[0, :], label="prey population", color="blue", alpha=0.5)
anx[0].plot(t, quantum[0, :], linestyle="dashed", color="blue")
anx[0].plot(t, numerical[1, :], label="predator population", color="red", alpha=0.5)
anx[0].plot(t, quantum[1, :], linestyle="dashed", color="red")
anx[0].set_ylim(np.min(numerical), np.max(numerical) * 1.2)
anx[0].set_xlabel("time")
anx[0].set_ylabel("prey population")
anx[0].legend(loc="upper right")

anx[1].plot(numerical[0, :], numerical[1, :], label="numerical", color="black", alpha=0.5)
anx[1].plot(quantum[0, :], quantum[1, :], label="quantum", linestyle="dashed", color="black")
anx[1].set_xlabel("prey population")
anx[1].set_ylabel("predator population")
anx[1].legend(loc="upper right")
```

```python
# Load chemotherapy data
chemo_data_numerical = np.load("results/chemo-final_numerical.npz")
chemo_data_quantum = np.load("results/chemo-final_quantum.npz")
t_chemo = chemo_data_numerical["t"]
numerical_chemo = chemo_data_numerical["x"]
quantum_chemo = chemo_data_quantum["x"]

C_num = numerical_chemo[0,:]
C_q = quantum_chemo[0,:]

E_num = numerical_chemo[1,:]
E_q = quantum_chemo[1,:]

T_num = numerical_chemo[2,:]
T_q = quantum_chemo[2,:]
```

```python
# compare the quantum/numerical trajectories for PK component and PD component
fig, anx = plt.subplots(1, 2, figsize=(8, 3))
anx[0].plot(t_chemo, C_num, label="plasma concentration", color="blue", alpha=0.5)
anx[0].plot(t_chemo,C_q, linestyle="dashed", color="blue")
anx[0].plot(t_chemo, E_num, label="effect site concentration", color="red", alpha=0.5)
anx[0].plot(t_chemo, E_q, linestyle="dashed", color="red")

anx[0].set_xlabel("time (days)")
anx[0].set_ylabel("concentration")
anx[0].legend(loc="upper right")

anx[1].plot(t_chemo, T_num, label="numerical", color="black", alpha=0.5)
anx[1].plot(t_chemo, T_q, label="quantum", linestyle="dashed", color="black")
anx[1].set_xlabel("time (days)")
anx[1].set_ylabel("relative tumor size")
anx[1].legend(loc="upper left")
```

```python
# Load cytokine data
ck_data_numerical = np.load("results/ck-final_numerical.npz")
ck_data_quantum = np.load("results/ck-final_quantum.npz")

t_ck = ck_data_numerical["t"]
numerical_ck = ck_data_numerical["x"]
quantum_ck = ck_data_quantum["x"]

H_num = numerical_ck[0,:]
H_q = quantum_ck[0,:]

T_num = numerical_ck[1,:]
T_q = quantum_ck[1,:]

D_num = numerical_ck[2,:]
D_q = quantum_ck[2,:]

C_num = numerical_ck[3,:]
C_q = quantum_ck[3,:]
```

```python
# compare the quantum/numerical trajectories for PK component and PD component
fig, anx = plt.subplots(1, 2, figsize=(8, 3))
anx[0].plot(t_ck, D_num, label="drug concentration", color="blue", alpha=0.5)
anx[0].plot(t_ck, D_q, linestyle="dashed", color="blue")
anx[0].plot(t_ck, C_num, label="cytokine concentration", color="red", alpha=0.5)
anx[0].plot(t_ck, C_q, linestyle="dashed", color="red")

anx[0].set_ylim(0, np.max([np.max(D_num), np.max(C_num)]) * 1.2)
anx[0].set_xlabel("time (days)")
anx[0].set_ylabel("concentration")
anx[0].legend(loc="upper right")


anx[1].plot(t_ck, T_num, label="tumor", color="green", alpha=0.5)
anx[1].plot(t_ck, T_q, linestyle="dashed", color="green")
anx[1].plot(t_ck, H_num, label="healthy tissue", color="orange", alpha=0.5)
anx[1].plot(t_ck, H_q, linestyle="dashed", color="orange")
anx[1].set_ylim(0, np.max([np.max(T_num), np.max(H_num)]) * 1.2)

anx[1].set_xlabel("time (days)")
anx[1].set_ylabel("relative tissue size")
anx[1].legend(loc="upper right")
plt.savefig("plots/qva_qsp.png")
```
